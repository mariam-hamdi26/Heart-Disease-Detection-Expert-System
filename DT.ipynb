```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
from sklearn.model_selection import train_test_split 
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report
import joblib 
```

```python
df=pd.read_csv("D:/Semster 6/Intelligent Programming/HeartDisease/cleaned_data.csv")
df.head()
```

```python
x=df.drop("Heart_Disease_Target",axis=1)
y=df["Heart_Disease_Target"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model=DecisionTreeClassifier(random_state=42,min_samples_split=2,max_depth=None)
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
print("Classification Report:",classification_report(y_test,y_pred))
```

```python
joblib.dump(model, 'heart_disease_model.pkl')
```

```python
import collections
import collections.abc

collections.Mapping = collections.abc.Mapping
collections.MutableMapping = collections.abc.MutableMapping
collections.Sequence = collections.abc.Sequence

# Import Experta
from experta import *

# Define Patient Fact
class Patient(Fact):
    pass

# Expert System
class HeartDiseaseExpertSystem(KnowledgeEngine):

    def __init__(self):
        super().__init__()
        self.risk_score = 0

    # High Risk Rules
    @Rule(Patient(Cholesterol=P(lambda x: x > 240)))
    def high_cholesterol(self):
        self.risk_score += 2

    @Rule(Patient(Age=P(lambda x: x > 50)))
    def old_age(self):
        self.risk_score += 1

    @Rule(Patient(Resting_Blood_Pressure=P(lambda x: x > 140)))
    def high_bp(self):
        self.risk_score += 2

    @Rule(Patient(Maximum_Heart_Rate=P(lambda x: x < 100)))
    def low_hr(self):
        self.risk_score += 1

    @Rule(Patient(ST_Depression=P(lambda x: x > 2)))
    def st_depression_high(self):
        self.risk_score += 2

    @Rule(Patient(Exercise_Angina_Yes=1))
    def exercise_angina(self):
        self.risk_score += 2

    # Medium Risk Rules
    @Rule(Patient(Cholesterol=P(lambda x: 200 < x <= 240)))
    def medium_chol(self):
        self.risk_score += 1

    @Rule(Patient(Age=P(lambda x: 40 < x <= 50)))
    def medium_age(self):
        self.risk_score += 1

    @Rule(Patient(ST_Slope_Flat=1))
    def flat_slope(self):
        self.risk_score += 1

    # Low Risk Rules
    @Rule(Patient(Cholesterol=P(lambda x: x < 200)))
    def low_chol(self):
        self.risk_score += 0

    @Rule(Patient(Exercise_Angina_No=1))
    def no_angina(self):
        self.risk_score += 0

    @Rule(Patient(ST_Slope_Upsloping=1))
    def up_slope(self):
        self.risk_score += 0

    # Final Decision
    def get_result(self):
        if self.risk_score >= 6:
            return "High Risk"
        elif self.risk_score >= 3:
            return "Medium Risk"
        else:
            return "Low Risk"


# User Input
def get_user_input():
    age = int(input("Enter Age: "))
    chol = int(input("Enter Cholesterol: "))
    bp = int(input("Enter Resting Blood Pressure: "))
    hr = int(input("Enter Maximum Heart Rate: "))
    st = float(input("Enter ST Depression: "))
    angina = int(input("Exercise Angina? (1 Yes / 0 No): "))
    slope_flat = int(input("ST Slope Flat? (1 Yes / 0 No): "))
    slope_up = int(input("ST Slope Upsloping? (1 Yes / 0 No): "))

    return {
        "Age": age,
        "Cholesterol": chol,
        "Resting_Blood_Pressure": bp,
        "Maximum_Heart_Rate": hr,
        "ST_Depression": st,
        "Exercise_Angina_Yes": angina,
        "Exercise_Angina_No": 1 - angina,
        "ST_Slope_Flat": slope_flat,
        "ST_Slope_Upsloping": slope_up
    }


#Run System
if __name__ == "__main__":
    engine = HeartDiseaseExpertSystem()

    user_data = get_user_input()

    engine.reset()
    engine.declare(Patient(**user_data))
    engine.run()

    print("\nFinal Prediction:", engine.get_result())
```

```python
print("Risk Score:", engine.risk_score)
print("Final Prediction:", engine.get_result())
```
